```python
# ==============================================================================
# ACTIVIDAD 2: ANÁLISIS DE MARKETING VIRAL CON APACHE SPARK Y CRISP-ML
# CURSO: Ingeniería de Software con IA - SENATI
# ==============================================================================

# ------------------------------------------------------------------------------
# FASE 1: INSTALACIÓN Y MONTAJE DE DRIVE
# ------------------------------------------------------------------------------
from google.colab import drive
drive.mount('/content/drive')

!pip install -q pyspark openpyxl pandas

import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, lower, regexp_replace
from pyspark.ml.feature import (
    Tokenizer, StopWordsRemover, HashingTF, IDF,
    StringIndexer, OneHotEncoder, VectorAssembler
)
from pyspark.ml.classification import LogisticRegression
from pyspark.ml import Pipeline
from pyspark.ml.evaluation import BinaryClassificationEvaluator, MulticlassClassificationEvaluator

# Inicializar SparkSession
spark = SparkSession.builder \
    .appName("CRISP_ML_Marketing_Viral") \
    .getOrCreate()

print("✓ Entorno configurado y SparkSession lista.")

# ------------------------------------------------------------------------------
# FASE 2: COMPRENSIÓN Y CARGA DE DATOS (Data Understanding)
# ------------------------------------------------------------------------------
# Cargar el Excel desde tu Google Drive
ruta_drive = '/content/drive/MyDrive/dataset_marketing_viral.xlsx'
df_pandas = pd.read_excel(ruta_drive)
df_spark = spark.createDataFrame(df_pandas)

print("\n--- Esquema del Dataset Cargado ---")
df_spark.printSchema()

print("\n--- Vista de los primeros 5 registros ---")
df_spark.show(5, truncate=False)

# Total de registros
print("Total de registros cargados:", df_spark.count())

# ------------------------------------------------------------------------------
# FASE 3: PREPARACIÓN DE DATOS (Data Preparation)
# ------------------------------------------------------------------------------
# 1. Limpieza de texto en 'texto_post' (convertir a minúsculas y quitar caracteres especiales)
df_prep = df_spark.withColumn("texto_limpio", lower(col("texto_post")))
df_prep = df_prep.withColumn("texto_limpio", regexp_replace(col("texto_limpio"), "[^a-zA-záéíóúÁÉÍÓÚñÑ0-9 ]", ""))

# 2. Configurar etapas del Pipeline de Feature Engineering
tokenizer = Tokenizer(inputCol="texto_limpio", outputCol="words")
remover = StopWordsRemover(inputCol="words", outputCol="filtered_words")
remover.setStopWords(StopWordsRemover.loadDefaultStopWords("spanish"))
hashingTF = HashingTF(inputCol="filtered_words", outputCol="raw_text_features", numFeatures=100)
idf = IDF(inputCol="raw_text_features", outputCol="text_features")

# Encoding para variable categórica 'contenido_multimedia'
indexer = StringIndexer(inputCol="contenido_multimedia", outputCol="multimedia_index")
encoder = OneHotEncoder(inputCol="multimedia_index", outputCol="multimedia_vec")

# VectorAssembler: Unir todas las características (texto + multimedia + números)
assembler = VectorAssembler(
    inputCols=["text_features", "multimedia_vec", "hora_publicacion", "num_seguidores_autor"],
    outputCol="features"
)

# ------------------------------------------------------------------------------
# FASE 4: MODELADO CON MLLIB (Modeling)
# ------------------------------------------------------------------------------
# Separar dataset: 80% Entrenamiento, 20% Prueba
train_data, test_data = df_prep.randomSplit([0.8, 0.2], seed=42)

# Algoritmo de clasificación
lr = LogisticRegression(labelCol="target", featuresCol="features", maxIter=20)

# Ensamblar Pipeline completo de Spark
pipeline = Pipeline(stages=[
    tokenizer, remover, hashingTF, idf,
    indexer, encoder, assembler, lr
])

# Entrenar el modelo
modelo_viralidad = pipeline.fit(train_data)
print("\n✓ Modelo de Machine Learning entrenado exitosamente con Spark MLlib.")

# ------------------------------------------------------------------------------
# FASE 5: EVALUACIÓN DEL MODELO (Evaluation)
# ------------------------------------------------------------------------------
# Predicciones sobre el conjunto de test
predictions = modelo_viralidad.transform(test_data)

print("\n--- Muestra de Predicciones sobre Datos de Test ---")
predictions.select("texto_post", "contenido_multimedia", "target", "prediction").show(5, truncate=False)

# Métricas
evaluator_roc = BinaryClassificationEvaluator(labelCol="target", metricName="areaUnderROC")
evaluator_acc = MulticlassClassificationEvaluator(labelCol="target", predictionCol="prediction", metricName="accuracy")

auc = evaluator_roc.evaluate(predictions)
acc = evaluator_acc.evaluate(predictions)

print(f"\n📊 Área Bajo la Curva ROC (AUC-ROC): {auc:.4f}")
print(f"📊 Precisión del Modelo (Accuracy):  {acc * 100:.2f}%")

# ------------------------------------------------------------------------------
# FASE 6: PRUEBA DE DESPLIEGUE / INFERENCIA (Deployment Simulation)
# ------------------------------------------------------------------------------
# Probar el modelo con nuevas publicaciones antes de publicarlas
nuevos_posts = spark.createDataFrame([
    ("¡Increíble unboxing del nuevo TecnoNova X1! Miren todas sus características en este video", 20, 50000, "Video"),
    ("Horario de atención presencial de TecnoNova por feriado", 8, 1500, "Solo texto")
], ["texto_post", "hora_publicacion", "num_seguidores_autor", "contenido_multimedia"])

# Preprocesar e inferir
nuevos_posts = nuevos_posts.withColumn("texto_limpio", regexp_replace(lower(col("texto_post")), "[^a-zA-záéíóúÁÉÍÓÚñÑ0-9 ]", ""))
resultados_inferencia = modelo_viralidad.transform(nuevos_posts)

print("\n--- Resultados de Predicción para Nuevos Posts ---")
resultados_inferencia.select(
    col("texto_post").alias("Post"),
    col("contenido_multimedia").alias("Tipo"),
    col("prediction").alias("Predicción (1=Viral, 0=No Viral)")
).show(truncate=False)
```
